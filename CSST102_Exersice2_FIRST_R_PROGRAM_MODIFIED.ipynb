**11. Creating a Simple Rule-Based System**

```r
classify_risk <- function(attendance, grade)
if (attendance <= 75 && grade <= 75) {
  print("High Risk")
} else {
  "Low Risk"
}

classify_risk(84, 95)
```

**12. Add More Rules**

```r
student_risk <- function(attendance, grade)
if (attendance < 75 && grade < 75) {
  print("High Risk")
} else if (attendance < 85 || grade < 80) {
  print("Moderate Risk")
} else if (attendance >= 85 && grade >= 80) {
  "Low Risk"
}

student_risk(96, 75)
```

**13. Apply the Rule to Multiple Students**

```r
students$Risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
)

students
```

**15. Add a Recomendation**

```r
recommend_Action <- function(risk_Level)
if (risk_Level == "High Risk") {
  return ("Immediate academic intervention")
} else if (risk_Level == "Moderate Risk") {
  return ("Academic Monitoring")
} else {
  return ("Continue Regular Monitoring")
}

students$recommendation <- sapply(students$Risk, recommend_Action)

students
```

**16. Mini KRR System (Final System)**

```r
student_id <- "S001"
attendance <- 90
grade <- 80

if (attendance < 75 && grade < 75) {
  risk <- "High Risk"
} else if (attendance < 85 || grade < 80) {
  risk <- "Moderate Risk"
} else {
  risk <- "Low Risk"
}

get_recommendation <- function(risk) {
  if (risk == "High Risk") {
    return("Immediate academic intervention")

  } else if (risk == "Moderate Risk") {
    return("Academic monitoring")

  } else {
    return("Continue regular monitoring")
  }
}

recommendation <- get_recommendation(risk)
cat("Student ID:", student_id, "\n")
cat("Attendance:", attendance, "%\n")
cat("Grade:", grade, "\n\n")
cat("Risk Level:", risk, "\n\n")
cat("Recommendation:\n")
cat(recommendation, "\n")
```
