# **7. Implement the Rules**

**Rule 1 -** *Eligible*: Programming Grade >= 80 AND Math Grade >= 75 AND Attendance >= 80% AND Prerequisite = Yes.

**Rule 2 -** *Conditional Eligible*:  Programming Grade >= 75 AND Math Grade >= 70 AND Attendance >= 75%, but Rule 1 is not satisfied.

**Rule 3 -** *Not Eligible*: If neither Rule 1 nor Rule 2 is satisfied, classify the student as Not Eligible.

```r
check_eligibility <- function(programming, math, attendance, prereq) {

  if (programming >= 80 &&
      math >= 75 &&
      attendance >= 80 &&
      prereq == "Yes") {
    return("Eligible")

  } else if (programming >= 75 &&
             math >= 70 &&
             attendance >= 75) {
    return("Conditional Eligible")

  } else {
    return("Not Eligible")

  }
}

eligibility <- check_eligibility(
  programming,
  math,
  attendance,
  prereq
)

cat("Eligibility Status:", eligibility)
```

# **8. Apply the Function to All Students**

```r
students$Eligibility <- mapply(
  check_eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

# **9. Create Recommendation Rules**

  **1. Eligible =** Student may Enroll
  
  **2. Conditional Eligible =** Consult Academic Adviser

  **3. Not Eligible =** Complete requirements before enrollment

```r
get_recommendation <- function(eligibility) {

  if (eligibility == "Eligible") {
    return("Student may Enroll")

  } else if (eligibility == "Conditional Eligible") {
    return("Consult Academic Adviser")

  } else {
    return("Complete requirements before enrollment")
  }
}

students$Recommendation <- sapply(
  students$Eligibility,
  get_recommendation
)

students
```
